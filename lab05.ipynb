## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette="hsv")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: The y-axis starts at 3,000 instead of 0, which makes the difference in body mass between the penguin species look larger than they actually are. The hsv color palette also makes the plot somewhatt distracting.


How you changed the plot and why: Yeah, we changed the y-axis to start at 0 instead of 3,000. This makes the differences between the species more accurate and prevents the plot from exaggerating the differences.
