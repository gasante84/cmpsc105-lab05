## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette="hsv")
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: The y-axis starts at 3,000 instead of 0, which makes the difference in body mass between the penguin species look larger than they actually are. The hsv color palette also makes the plot somewhatt distracting.


How you changed the plot and why: Yeah, we changed the y-axis to start at 0 instead of 3,000. This makes the differences between the species more accurate and prevents the plot from exaggerating the differences.


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds = sns.load_dataset("diamonds")

sns.scatterplot(data=diamonds, x="carat", y="price", color="blue", s=20, alpha=0.3)
plt.title("Diamond Price vs. Carat")
plt.show()
```

Issues with the plot: The plot has a very large number of points, and the points are large and overlap each other. This creates a solid block of ink and makes it difficult to see the distribution of the data.


How you changed the plot and why: We made the points smaller and more transparent. This reduces the amount of overlap and makes it easier to see where the data are concentrated.


## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

sns.lineplot(data=flights, x="month", y="passengers", units="year", estimator=None, hue="year", palette="viridis")
plt.title("Passengers over Time")
plt.show()
```

Issues with the plot: There are many lines on the graph, and they are all black, which makes it difficult to distinguish between the different years. The lines overlap and make it difficult to follow the passenger numbers for individual years.


How you changed the plot and why: We changed the lines from color black to different colors for each year and added a legend. This makes it easier to distinguish between the years and compare passenger numbers over time.


## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker", scatter_kws={"s": 30, "alpha": 0.4})
plt.show()
```

Issues with the plot: We could see a heavy overlap of the points, especially where most of the observations are concentrated. The two smoking groups are also shown on the same graph, which makes it more difficult to see and compare the relationships between total bill and tip.


How you changed the plot and why: We made the points smaller and more transparent to reduce the overlap. We also kept the smoking groups separated by using the existing hue="smoker" variable, so we could still compare the two groups.
